**TAKE-HOME EXERCISE**

Complete the following exercise for the next session:

You would like to get a male puppy that suits your lifestyle. To help you with your decision you obtain information on 97 dog breeds (captured in a table in the file dog_breeds.csv). According to the data in the table, which dog breeds can you consider?

The dog should:

Have a below average height (take the maximum height of the for male dogs as the expected height for a breed)

Definitely not be a barker (barking scores of 1 or 2 only)

Be good with children (good_with_children score of 4 or 5)

```python
import pandas as pd

# Load the dataset
df = pd.read_csv('dog_breeds.csv')

# Calculate the average of the maximum height for male dogs
avg_max_height_male = df['max_height_male'].mean()

# Filter based on the specified criteria
# 1. Below average maximum height for male dogs
# 2. Barking score of 1 or 2 only
# 3. Good with children score of 4 or 5
matching_breeds = df[
    (df['max_height_male'] < avg_max_height_male) &
    (df['barking'].isin([1, 2])) &
    (df['good_with_children'].isin([4, 5]))
]

print(f"Average maximum height for male dogs: {avg_max_height_male:.2f} inches\n")
print("Dog breeds that fit your criteria:")
print(matching_breeds[['Name', 'max_height_male', 'barking', 'good_with_children']])
```
